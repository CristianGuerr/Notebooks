# INTRODUCCIÓN

¡Claro, Lighty! Aquí tienes una versión breve, clara y actualizada (para 2026) de la introducción que puedes poner como primera celda de texto (Markdown) en tu Google Colab:

```markdown
## Tabla de Contenidos Interactiva en Google Colab

En Google Colab puedes crear una **tabla de contenidos automática e interactiva** que permite navegar fácilmente entre las diferentes secciones del notebook con un solo clic.

### Método principal (recomendado – nativo y automático)
1. Usa encabezados en celdas de texto Markdown con `#`, `##`, `###`, etc.:
   - `# Título principal`
   - `## Subsección`
   - `### Apartado más pequeño`
   
   Colab genera automáticamente los enlaces internos.

2. Activa la tabla de contenidos:
   - Ve al menú superior → **Ver** → **Mostrar índice**  
     (o usa el atajo **Ctrl + Shift + I** en la mayoría de los casos)
   
   → Aparecerá un panel lateral a la izquierda con todos los encabezados.  
   Puedes hacer clic en cualquier entrada para saltar directamente a esa sección.  
   El índice se actualiza solo cada vez que agregas/modificas encabezados.

### Ventajas
- No requiere código extra
- Funciona en cualquier notebook (incluso muy largos)
- Es plegable y se mantiene visible mientras navegas
- Ideal para reportes, tutoriales o análisis extensos

¡Con solo estructurar bien tus títulos ya tienes navegación profesional en segundos! 🚀

(Opcional: si prefieres tener la tabla **dentro** del flujo del notebook en vez de en la barra lateral, hay trucos con HTML/JS o extensiones, pero el método nativo es el más simple y confiable en 2026.)
```

# Fundamento teórico
**Fundamentos teóricos básicos de la navegación por secciones en Google Colab**

- Los encabezados Markdown (#, ##, ###…) se convierten en elementos HTML `<h1>`, `<h2>`, `<h3>` con IDs automáticos (slugificados).
- Estos IDs funcionan como **anclas HTML** (#mi-seccion).
- Al hacer clic en un enlace con #id, el navegador realiza **scroll suave** al elemento correspondiente (comportamiento nativo de HTML5).
- Colab analiza dinámicamente todos los encabezados → genera un **índice jerárquico** (TOC) en la barra lateral izquierda con enlaces que usan esos mismos anclas.
- Todo ocurre en el cliente (JavaScript del navegador), sin recargar la página.

En una frase: **Markdown → anchors HTML → scroll nativo del navegador + TOC dinámico generado por Colab**.

# EJEMPLO PRÁCTICO
}La fórmula de Ramanujan (una de las más conocidas para calcular π) es:

1π=22–√9801∑k=0∞(4k)!(1103+26390k)(k!)43964k

Esta serie converge extremadamente rápido, generando muchos dígitos de precisión con pocos términos.

import math
from decimal import Decimal, getcontext


def factorial(n):
    """Calcula el factorial de n"""
    if n == 0 or n == 1:
        return 1
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result


def ramanujan_pi(terminos=10, precision=50):
    """
    Calcula π usando la serie de Ramanujan (la más conocida)
    1/π = (2√2 / 9801) * Σ [ (4k)! * (1103 + 26390k) / ( (k!)^4 * 396^(4k) ) ]
    
    Parámetros:
    - terminos: cuántos términos de la serie usar (1 ya da ~8 dígitos, 5 dan ~40 dígitos)
    - precision: cuántos dígitos decimales de precisión queremos
    
    Retorna: π como Decimal con alta precisión
    """
    # Configuramos precisión decimal alta
    getcontext().prec = precision + 10  # un poco más para evitar errores de redondeo
    
    C = Decimal(396)           # 396
    A = Decimal(2) * Decimal(2).sqrt() / Decimal(9801)   # 2√2 / 9801
    suma = Decimal(0)
    
    for k in range(terminos):
        num_fact4k = Decimal(factorial(4 * k))
        den_factk4 = Decimal(factorial(k)) ** 4
        numerador_lineal = Decimal(1103 + 26390 * k)
        denominador_potencia = C ** (4 * k)
        
        termino = num_fact4k * numerador_lineal / (den_factk4 * denominador_potencia)
        suma += termino
        
        # Opcional: mostrar convergencia por término
        # print(f"k={k:2d} → término = {termino:.15f}")
    
    uno_sobre_pi = A * suma
    pi = 1 / uno_sobre_pi
    
    return pi


# Ejemplo de uso
if __name__ == "__main__":
    # Queremos 40 dígitos de precisión y usamos 6 términos
    getcontext().prec = 45
    pi_calculado = ramanujan_pi(terminos=6, precision=40)
    
    print("π según Ramanujan (6 términos) ≈")
    print(pi_calculado)
    
    print("\nComparación con math.pi:")
    print(math.pi)
    
    # Para ver más dígitos
    print("\nCon más precisión (primeros 40 dígitos):")
    print(str(pi_calculado)[:42])
