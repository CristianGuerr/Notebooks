```python
# ENTRADA:
base = float(input("Ingrese la base en metros:"))
altura = float(input("Ingrese la altura en metros:"))
longitud = float(input("Ingrese la longitud en metros:"))

# PROCESO:
densidad = 7850
peso = (densidad * base * altura * longitud)

# SALIDA:
print("El peso de la viga es:", peso, "kg")
```
