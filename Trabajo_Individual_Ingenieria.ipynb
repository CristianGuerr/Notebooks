```python
# ENTRADAS: Bultos de cemento se necesitan para un muro de ladrillo
area = (input("Ingrese el área del muro en m²: "))
area = float(area)

# PROCESO: Calcular cuántos bultos de cemento se necesitan para un muro de ladrillo en m^2
bultos = (area * 0.25)

# SALIDA:
print("Se necesitan", bultos, "bultos de cemento")
```
